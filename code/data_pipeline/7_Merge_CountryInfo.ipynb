```python
# First authorship 




use_col = ['index_num','country','Publication Year','affiliation']

first = pd.read_csv(FIRST_IN, usecols=use_col)


# --- 1. Set country to NaN if affiliation is NaN ---
first.loc[first['affiliation'].isna(), 'country'] = np.nan

# --- 2. Work only with rows that have affiliation and country ---
first_nonan = first[first['affiliation'].notna() & first['country'].notna()]

# --- 3. Count frequency of each country per affiliation ---
aff_country_counts = first_nonan.groupby(['affiliation', 'country']).size().reset_index(name='count')

# --- 4. Find max count per affiliation ---
max_counts = aff_country_counts.groupby('affiliation')['count'].max().reset_index(name='max_count')

# --- 5. Merge to identify countries with max frequency ---
aff_country_max = aff_country_counts.merge(max_counts, on='affiliation')
aff_country_max = aff_country_max[aff_country_max['count'] == aff_country_max['max_count']]

# --- 6. Identify affiliations with ties ---
tie_affiliations = aff_country_max.groupby('affiliation').size()
tie_affiliations = tie_affiliations[tie_affiliations > 1].index

# --- 7. Keep only affiliations without tie for canonical mapping ---
canonical_mapping = aff_country_max[~aff_country_max['affiliation'].isin(tie_affiliations)]
aff_country_dict = canonical_mapping.set_index('affiliation')['country'].to_dict()

# --- 8. Map canonical country to all rows, keep original if tie or NaN ---
first['country_ins'] = first['affiliation'].map(aff_country_dict).combine_first(first['country'])

# --- 9. Clean up to free memory ---
del first_nonan, aff_country_counts, max_counts, aff_country_max, canonical_mapping, aff_country_dict, tie_affiliations
```
